```python
import numpy as np
print(np.pi)
```

```python
pi_expansion = sum([4 * (-1)**k / (2*k + 1) for k in range(1000000)])
print(pi_expansion)
```

```python
random_points = np.random.rand(1000000, 2)
norms = np.linalg.norm(random_points, axis=1)
inside_circle = np.sum(norms <= 1)
total_points = random_points.shape[0]
pi_estimate = inside_circle / total_points
print(pi_estimate * 4)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 6))
expansion_terms = [4 * (-1)**k / (2*k + 1) for k in range(100)]
cumulative_sum = np.cumsum(expansion_terms)
plt.plot(cumulative_sum, label='Series')
```

```python
monte_carlo_cumsum = []
for i in range(1, 100001):
    random_points = np.random.rand(i, 2)
    norms = np.linalg.norm(random_points, axis=1)
    inside_circle = np.sum(norms <= 1)
    total_points = random_points.shape[0]
    pi_estimate = inside_circle / total_points
    monte_carlo_cumsum.append(pi_estimate * 4)
    
plt.plot(monte_carlo_cumsum, label='Monte Carlo')
```
